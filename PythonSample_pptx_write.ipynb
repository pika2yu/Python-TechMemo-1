```python
import pptx
```

```python
pptout = pptx.Presentation()
```

```python
#
# Shape の削除
#
# python-pptx の z-order を変えるための issue が参考。
# remove までで止めると shape が削除される。
#   https://github.com/scanny/python-pptx/issues/49
#
slide_index = 3
sld = pptin.slides[slide_index]

shape = sld.shapes[1]
print(shape.text)

spTree = sld.shapes._spTree
spTree.remove(shape._element)  # これで削除される
#spTree.insert(3, shape._element)

pptin.save('new.pptx')
```

```python
#
# slide の削除
# xml の slide list から、削除すると消える。
# 参考URL:
#   https://stackoverflow.com/questions/54945022/unable-to-delete-powerpoint-slides-using-python-pptx
#
pptin = pptx.Presentation('sample1.pptx')

slide_index = 3

xml_slides = pptin.slides._sldIdLst  
slides = list(xml_slides)
xml_slides.remove(slides[slide_index])

pptin.save('new.pptx')
```

```python
#
# Slide の挿入
# TBD. ootx を開く時、エラーが出る。何か足りない。
#
slide_index = 3
sld = pptin.slides[slide_index]

title_slide_layout = pptout.slide_layouts[0]
slide1 = pptout.slides.add_slide(title_slide_layout)
title = slide1.shapes.title
title.text = "Pythonで作成したタイトル"

pptout.slides.add_slide(sld.slide_layout)

pptout.save("new.pptx")
```
