```python
#
# 通常のインスタンス変数 - インスタンス毎に異なる値を持つ
# コンストラクタで定義するという以外は他の言語と同じ。
# (1). 変数はコンストラクタ (__init__(self) の中で定義する
# (2). クラス内では　self.変数名　でアクセスする
# (3). 外からのアクセスは インスタンス名.変数
#      クラス名.変数名　はエラー
#
class MyCounterInst:
    def __init__(self):
        self.cnt_instance = 0
    def increment(self):
        self.cnt_instance+=1

c1 = MyCounterInst()
c2 = MyCounterInst()

#
# クラス名.変数 でも インスタンス名.変数 アクセスできる
#
print('クラス名.変数 / インスタンス名.変数 でアクセス')
#print(f'{MyCounterInst.cnt_instance} ({id(MyCounter.cnt_instance)})')  # クラス名.変数名 はエラー
print(f'{c1.cnt_instance} ({id(c1.cnt_instance)})')
print(f'{c2.cnt_instance} ({id(c2.cnt_instance)})')
print()

#
# 書き換えは直接でもメソッド経由でもインスタンス毎に反映される
#
c1.cnt_instance+=1
print(f'{c1.cnt_instance} ({id(c1.cnt_instance)})')
print(f'{c2.cnt_instance} ({id(c2.cnt_instance)})')
print()

c1.increment()
print(f'{c1.cnt_instance} ({id(c1.cnt_instance)})')
print(f'{c2.cnt_instance} ({id(c2.cnt_instance)})')
print()
```

```python
#
# クラス変数 - 全てのinstanceに共通になる変数
# クラス変数はあまり使わないが、Python ではちょっとややこしい。
#
# (1). classのbodyで定義する
# (2). 読み込みは クラス名.変数 または インスタンス名.変数 で行える
# (3). 書き換える時は クラス名.変数 で行う
# (4). **NG** 書き換えを インスタンス名.変数 で行ってはいけない！
#      エラーにはならないが、そのインスタンス固有の同じ名前の変数が生成されてしまう。
# (5). クラスのインスタンスメソッドからアクセスする時は クラス名.変数
#      クラス名.メソッド名　で呼び出す時はパラメータにインスタンスをつけ加える。
#        例えば、MyCounter.increment2(c1)
#        実は、c1.increment2() は上記の省略形。
#        呼び出される側のメソッドで付ける第一引数の self は、このパラメータ(=インスタンス)の事。
# (6). @classmethod をつけるとクラスメソッド(=スタティクメソッド)になる。
#      クラスメソッドからアクセスする時は cls.変数。cls はメソッドの第一引数。
#      クラスメソッドは クラス名.メソッド名 でも インスタンス名.メソッド名 でも呼び出せる。
#
class MyCounter:
    cnt_class = 0                 # body で定義する
    @classmethod                  #
    def increment(cls):           # @classmethod をつけているのでクラスメソッド
        cls.cnt_class+=1          #   cls.変数名 (cls.cnt_class) でクラス変数を書き換える
    def increment2(self):         # インスタンスメソッド
        MyCounter.cnt_class+=1    #   クラス名.変数名 (MyCounter,cnt_class)) でクラス変数を書き換える

c1 = MyCounter()
c2 = MyCounter()

#
# クラス名.変数 でも インスタンス名.変数 アクセスできる
#
print('クラス名.変数 / インスタンス名.変数 でアクセス')
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

# 書き換える時は クラス名.変数 で行う
# 書き換えると id が変わるが、どのインスタンスでも id は同じ。
print('クラス名.変数 で書き換え。書き換えると id が変わるが、どのインスタンスでも id は同じ。')
MyCounter.cnt_class+=1
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

# クラスメソッド(@classmethodで定義)で書き換え
print('クラスメソッド(@classmethodで定義)で書き換え。クラス名.メソッド名で呼び出し。')
MyCounter.increment()
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

print('クラスメソッド(@classmethodで定義)で書き換え。インスタンス名.メソッド名で呼び出し。')
c1.increment()
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

# インスタンスメソッドで書き換え(内部でクラス名.変数 を使う)
print('インスタンスメソッドで書き換え(内部でクラス名.変数 を使う）')
c1.increment2()
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

# インスタンスメソッドで書き換え(内部でクラス名.変数 を使う)
print('インスタンスメソッドで書き換え(内部でクラス名.変数 を使う）')
MyCounter.increment2(c1)
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

#
# インスタンス名.変数 でアクセスすると、その名前の変数が新たに作られてしまいクラス変数へのアクセスにならない
# 同じ書き方(c1.cnt_class)で違うものを指すのが紛らわしい！！
#
print('**NGな使い方** インスタンス名.変数 で書き換え。書き換えたインスタンスのidが他とは違っている。')
c1.cnt_class+=1  # クラス変数の値に 1 を足した新しい変数になる
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')   # <<< c1に新しいローカルな変数ができる。
                                                #     同じ書き方(c1.cnt_class)で違うものを指すのが紛らわしい！！
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()

# 
# もう一度、正しく書き換えると、クラス名.変数名(c1.cnt_class)の参照先が元に戻るところが、また、ややこしい。
#
print('もう一度、正しく書き換えると、クラス名.変数名の参照先が元に戻るところが、また、ややこしい。')
# 書き換え方は、どれでも良い。
#MyCounter.increment()
c1.increment2()
#MyCounter.cnt_class+=1
print(f'{MyCounter.cnt_class} ({id(MyCounter.cnt_class)})')
print(f'{c1.cnt_class} ({id(c1.cnt_class)})')   # <<< c1 に新しいローカルな変数ができてしまう。
print(f'{c2.cnt_class} ({id(c2.cnt_class)})')
print()
```

```python
#
# クラス変数は クラス名.変数名 で使うか、メソッドで隠してしまうのが良いのだが、
# リストなどだと、間違った使い方でもそこそこ動いてしまい意図してない動きなことに気づかないのが問題！！
# 関数を使ってアクセスすると参照送りされるのでインスタンスローカルな変数は作られない。
#
class Test:
    a = []  # クラス変数

    def append(self, value):
        self.a.append(value)

    def clear(self):
        #self.a = []
        self.a.clear()

    def geta(self):
        return self.a

t1 = Test()
t2 = Test()
print(t1.a)
print(t2.a)

t1.append('123')
print(t1.a)
print(t2.a)

t1.append('')
print(t1.a)
print(t2.a)

t1.clear()
print(t1.a)
print(t2.a)
print(t1.geta())
print(t2.geta())

t1.append('456')
print(t1.a)
print(t2.a)

t1.a.clear()
print(t1.a)
print(t2.a)

t1.append('456')
print(t1.a)
print(t2.a)

t1.a=[]          # 変数を直接書き換えるとインスタンス独自のローカルな値になる。
print(t1.a)
print(t2.a)
```
