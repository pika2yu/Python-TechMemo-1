```python
#
# enumerate
# enumerate()を使うとforループ内でリストやタプルなどの要素と同時にインデックス（0, 1, 2...）を取得できる
#
fruites = ['apple', 'banana', 'orange', 'grapes', 'mango']

for idx, fr in enumerate(fruites):
    print(f'{idx},{fr}')

print('index のスタート値を start=X で指定する事も可能')
for idx, fr in enumerate(fruites, start=2):
    print(f'{idx},{fr}')

#
# ただのループだったら以下のように range を使う
#
print('ただのループ')
num_of_loop = 5
for icnt in range(num_of_loop):
    print(f'icnt={icnt}')
```

```python
#
# CSV の読み込み
#
import csv
fname = 'sample.csv'
with open(fname,'r', encoding='utf-8') as fl:
    reader = csv.reader(fl) # reader で1行づつ読み込める
    header = next(reader)   # header読み込み : 1行目がヘッダなら next で最初の行を読み込む
    print(header)
    for row in reader:      # 1行読み込んでいるので2行目からの繰り返し
        print(row)          # 読み込んだ行の要素(,で区切られたデータ)はリストになっている
        for cell in row:    # 各データを個別にアクセスできる
            print(cell)
```

```python
#
# Unicodeエスケープ関連の変換
#
print('***Unicodeエスケープから文字列への変換***')
cssUnicodeEscaped_str = '\\30c6\\30b9\\30c8'    # CSSで使うUnicodeエスケープ。/u が無い。
                                                # 'テスト' が  \30c6\30b9\30c8 にエスケープされている
print(f'元データ(css format) = {cssUnicodeEscaped_str}')
unicodeEscaped_str = cssUnicodeEscaped_str.replace('\\','\\u')   # /u をつけると通常の Unicodeエスケープになる。
print(f'元データ(unicode escape) = {cssUnicodeEscaped_str}')
# ここまでは エスケープした表現の文字列。

# エスケープをデコードして文字列に変換する時は、まず、エスケープしている文字列をutf-8でのバイナリに変換
utf8_UnicodeEscaped_bin = unicodeEscaped_str.encode('utf-8')
print(f'utf8 binary = {utf8_UnicodeEscaped_bin}')
# utf-8 のバイト列にした uinicode escapeデータを decode すると文字列になる。 
decodedStr=utf8_UnicodeEscaped_str.decode('unicode-escape') # unicode-escape表現のデータである事を明示する！
print(f'decoded = {decodedStr}')
# unicode-escape を明示していないと、unicode escapeの文字列に戻るだけ。
print(f'unicode-escape を指定しないで decode した時：{utf8_UnicodeEscaped_str.decode()}')


#1行で書くとこうなる
decodedStr=unicodeEscaped_str.encode('utf-8').decode('unicode-escape') # 文字に変換
print(decodedStr)

print()
#
# 逆向き
#
print('***逆向きの変換***')
text = 'テスト'
utf8_encoded_bin = text.encode('unicode-escape')
print(utf8_encoded_bin)
utf8_encoded_text = utf8_encoded_bin.decode()
print(utf8_encoded_text)
print(f'デフォルト(utf-8)でencodeするとunicodeのutf8表現 : {text.encode()}')
```

```python
#
# 端数は小数になる
#
print(10/6)

#
# 切り捨てて整数にする
#
print(10//6)

#
# round は四捨五入と説明と説明される時があるが。違う！！
# x.5 の時は近い方の偶数になる
#
print(round(10/6))
print(round(3/2)) # 1.5 なので 2。一見、四捨五入だが、、、
print(round(5/2)) # 近い方の偶数なので、2.5 は 3 ではなく 2。つまり、四捨五入では無い。
```
